```python
import pandas as pd
from collections import Counter
import re

# Load the dataset
df = pd.read_csv('/content/merged_output.csv')

# Combine text from all columns, handling null values
df['combined_text'] = df[['input', 'instruction', 'output']].fillna('').agg(' '.join, axis=1)

# Function to tokenize text (basic tokenization for Bangla or any language)
def tokenize(text):
    # Remove punctuation and split into words
    words = re.findall(r'\b\w+\b', text)
    return words

# Tokenize all text and count word frequencies
all_words = []
for text in df['combined_text']:
    all_words.extend(tokenize(text.lower()))  # Convert to lowercase for consistency

# Count frequencies
word_counts = Counter(all_words)

# Get the top 200 most common words
top_200_words = word_counts.most_common(200)

# Print the top 200 words with their frequencies
print("Top 200 Most Frequent Words:")
for word, freq in top_200_words:
    print(f"{word}: {freq}")
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# Create a dictionary from the top 200 words
word_freq_dict = dict(top_200_words)

# Generate the word cloud
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white',
    font_path='/content/noto_bangla_font/NotoSansBengaliUI-Bold.ttf'  # Ensure this font path is correct for Bangla
).generate_from_frequencies(word_freq_dict)

# Display the word cloud
plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("🔠 Top 200 Frequent Words Bubble (Word Cloud)", fontsize=16)
plt.tight_layout()
plt.show()
```
